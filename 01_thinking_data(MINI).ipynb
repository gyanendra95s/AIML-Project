## Suppose I am doing internship in amazon and I am allocated with the task that I have to do the following task from the given data as:
- If data is inconsistent(mix type)->Make it consistent
- Handles the mixing values
- If there is any duplicate data then Deduplicate it
- What's avg. rating?
- What's % of users gave poor rating?
- ans so on....



```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)

    return data
    
```

```python
data=load_data("store_data.json")
print(data)
print(type(data))
```

```python
# CLean & Structure the data 
def clean_data(data): 
    # Rating of Product should br in numeric 
    text_to_num={"one": 1,"two": 2,"three": 3,"four": 4,"five": 5} 
    cleaned_data=[]
    unique_users=set()
    
    for user in data:
        # Clean Rating 
        raw_rating=str(user["rating"]).strip().lower() 
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]

        user["rating"]=raw_rating

        # Handle Missing Values
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None

        # Deduplication
        if(user["name"].strip() in unique_users):
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)
        
    return cleaned_data
```

```python
d=clean_data(data)
print(d)
```

```python
# To get meaningful insights from data
def get_insights(d):

    #avg_rating
    total_rating=0
    for user in d:
        total_rating += float(user["rating"])
    print(f"avg rating is {total_rating/len(d)}")

    #Percentage of users with Poor Rating
    poor_rating=0
    for user in d:
        if(float(user["rating"])<3.0):
            poor_rating+=1
    print(f"Percentage of users with poor rating={(poor_rating/len(d))*100}")
```

```python
get_insights(d)
```

```python
# get Recommendation
def get_recommendation(d):
    recommendations=[]

    for user in d:
        curr_recomm={}
        curr_recomm["name"]=user["name"]

        if(float(user["rating"])>=4):
            curr_recomm["brand"]="Apple"
        else:
            curr_recomm["brand"]="Samsung"

        recommendations.append(curr_recomm)

    return recommendations
```

```python
get_recommendation(d)
```
